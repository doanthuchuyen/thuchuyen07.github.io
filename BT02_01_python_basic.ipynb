```python
def show_info(id_num, **details):
    print('Mã NV: ', id_num)
    for key in details:
        print(f'Thuộc tính [{key}] có giá trị [{details[key]}]')

show_info('NV001', department = 'IT', salary = 5000)
show_info('NV002', title = 'Manager', level = 2)

def find_item(target, *item_list):
    exists = 0
    for idx in range(len(item_list)):
        if target == item_list[idx]:
            print(f'{target} được tìm thấy tại chỉ mục {idx}')
            exists = 1
    if exists == 0:
        print(f'{target} không có trong danh sách')

find_item(99, 11, 22, 33)
find_item(50, 40, 50, 60, 50, 70)

product = lambda num1, num2: num1 * num2
print("Kết quả nhân là : ", product( 5, 12 ))
```
