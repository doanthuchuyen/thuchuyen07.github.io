```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
N=50
x = np.random.rand(N)
y = np.random.rand(N)
r = np.random.rand(N)
scale_factor = 1000
sizes = (r**2) * scale_factor
```

```python
colors = np.random.rand(N)
fig, ax = plt.subplots()
scatter = ax.scatter(x, y, s=sizes, c=colors, cmap='viridis', alpha=0.6, edgecolors='w')
ax.set_title(f'Hien thi Ngau nhien N={N} Hinh tron')
ax.set_xlabel('Toa do x')
ax.set_ylabel('Toa do y')
ax.set_xlim(0, 1.2)
ax.set_ylim(0, 1.2)
ax.set_aspect('equal', adjustable='box')
cbar = fig.colorbar(scatter)
cbar.set_label('Gia tri Mau sac Ngau nhien (0 den 1)')
plt.show()
```
