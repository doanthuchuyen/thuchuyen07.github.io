```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris
import plotly.express as px # Sử dụng Plotly để tạo biểu đồ 3D tương tác
from IPython.display import display
```

```python
iris = load_iris(as_frame=True)
df = iris.frame
```

```python
df.columns = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'species_code']
```

```python
df['species_name'] = df['species_code'].map(lambda x: iris.target_names[x])

print("5 dòng đầu tiên của DataFrame đã được chuẩn bị:")
display(df.head())
```

```python
feature_x = 'sepal_length'
feature_y = 'sepal_width'
feature_z = 'petal_length'
color_by = 'species_name'
```

```python
fig = px.scatter_3d(
    df,
    x=feature_x,
    y=feature_y,
    z=feature_z,
    color=color_by, # Phân biệt màu sắc theo tên loài hoa
    symbol=color_by, # Phân biệt hình dạng điểm (symbol) theo tên loài hoa
    title=f'Trực quan hóa Dữ liệu Iris 3D: {feature_x}, {feature_y}, {feature_z}',
    height=600
)
```

```python
fig.update_layout(
    margin=dict(l=0, r=0, b=0, t=40),
    scene=dict(
        xaxis_title=feature_x.upper(),
        yaxis_title=feature_y.upper(),
        zaxis_title=feature_z.upper()
    )
)
```

```python
fig.show()

print("\nBiểu đồ 3D tương tác đã được tạo bằng Plotly.")
```
